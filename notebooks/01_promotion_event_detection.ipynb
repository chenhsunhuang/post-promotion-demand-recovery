# 01 | Promotion Event Detection

Identify product-level promotion periods from daily discount depth and demand peaks. The multi-stage algorithm uses a 14-day window, median absolute deviation (MAD), rolling thresholds and event-level filtering.

```python
from pathlib import Path
import sys
ROOT = Path.cwd().resolve()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
import pandas as pd
import numpy as np
from IPython.display import display
from src.promotion_detection import detect_promotions
```

## Demonstration with synthetic daily records

A fictional product illustrates the detection rules. The original transaction data are not included.

```python
dates = pd.date_range('2024-01-01', periods=80)
qty = np.full(80, 10.0)
discount = np.full(80, 1.0)
qty[35:40] = 35
discount[35:40] = 8
qty[40:44] = 4
synthetic = pd.DataFrame({'SalePageId': 'demo-product', 'OrderDate': dates,
                          'Qty': qty, 'SubtotalPromotionDiscount': -qty * discount,
                          'SubtotalCouponDiscount': 0.0, 'day_UnitPrice': 20.0,
                          'day_SubtotalSalesAmount': qty * 20})
marked = detect_promotions(synthetic)
display(marked[['is_promo_day_v1','is_promo_day_v2','is_promo_day_v3','Qty_peak_v2']].sum().rename('flagged_days'))
```
