# 02 | Demand Recovery Time (K)

Consolidate promotion days into events, identify a clean pre-promotion baseline, and compare four definitions of post-event recovery. K counts days from the end of the promotion to the day the recovery criterion is confirmed.

```python
from pathlib import Path
import sys
ROOT = Path.cwd().resolve()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
import pandas as pd
import numpy as np
from IPython.display import display
from src.recovery_time import (compute_k_cumulative_mean, compute_k_rolling_3day_mean,
    compute_k_consecutive_days, prepare_daily_data, calculate_all_k)
```

## Illustrative recovery sequence

```python
post_qty = [3, 5, 10, 12, 11, 9]
baseline = 10
pd.Series({'Cumulative mean': compute_k_cumulative_mean(post_qty, baseline),
           '3-day moving mean': compute_k_rolling_3day_mean(post_qty, baseline),
           '2 consecutive days': compute_k_consecutive_days(post_qty, baseline, 2),
           '3 consecutive days': compute_k_consecutive_days(post_qty, baseline, 3)})
```

## Reported comparison
